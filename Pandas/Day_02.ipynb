```python
# 6. Filter Customers by Name Pattern ....................................................................................
```

```python
# Starts with A
starts_A = Customer_Details[
    Customer_Details['customer_name'].str.contains(r'^A', case=False, na=False)
]

# Ends with n
ends_n = Customer_Details[
    Customer_Details['customer_name'].str.contains(r'n$', case=False, na=False)
]

print(starts_A)
print(ends_n)
```
